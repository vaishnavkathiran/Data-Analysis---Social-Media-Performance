```python
smp["day_type"] = np.where(
    smp["post_datetime"].dt.dayofweek >= 5,
    "Weekend",
    "Weekday"
)
```

```python
def categorize_sentiment(score):
    if score < -0.33:
        return "Negative"
    elif score <= 0.33:
        return "Neutral"
    else:
        return "Positive"

smp["sentiment_category"] = smp["sentiment_score"].apply(
    categorize_sentiment
)
```

```python
smp["hashtags"].head()
```

```python
smp["hashtag_count"] = (
    smp["hashtags"]
    .fillna("")
    .str.split()
    .str.len()
)
```

```python
platform_content = pd.pivot_table(
    smp,
    values="engagement_rate",
    index="platform",
    columns="content_type",
    aggfunc="mean"
)

platform_content
```

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    platform_content,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Average Engagement Rate by Platform and Content Type")
plt.xlabel("Content Type")
plt.ylabel("Platform")
plt.show()
```

```python
viral_platform = pd.crosstab(
    smp["platform"],
    smp["is_viral"]
)

viral_platform
```

```python
viral_platform.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Viral vs Non-Viral Posts by Platform")
plt.xlabel("Platform")
plt.ylabel("Number of Posts")
plt.xticks(rotation=45)
plt.legend(
    title="Viral Status",
    labels=["Non-Viral", "Viral"]
)
plt.show()
```

```python
correlation_columns = [
    "views",
    "likes",
    "comments",
    "shares",
    "engagement_rate",
    "sentiment_score"
]

correlation = smp[correlation_columns].corr()

correlation
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix of Social Media Performance Metrics")
plt.show()
```
